```python
#Load dataset
import pandas as pd

df= pd.read_csv(r"C:\MY DOCUMENTS\Saiket System Projects\Telco_Customer_Churn_Dataset  (3).csv")
print(df.head())
```

```python
#Standardize column names
df.columns = (df.columns
              .str.strip()          
              .str.lower()          
              .str.replace(' ', '_', regex=False))

print(list(df.columns))
```

```python
#Fix TotalCharges column
#convert to float
df['totalcharges'] = pd.to_numeric(df['totalcharges'], errors='coerce')

nan_count = df['totalcharges'].isnull().sum()
print(nan_count)

#Fill NaN with 0
df['totalcharges'] = df['totalcharges'].fillna(0.0)

print(df['totalcharges'].isnull().sum())
```

```python
#Change seniorcitizen column from (0, 1) to (No, Yes)

df['seniorcitizen']= df['seniorcitizen'].map({0:'No', 1:'Yes'})
print(df['seniorcitizen'].value_counts().to_string())
```

```python
#Remove duplicate rows
before = len(df)
df.drop_duplicates(inplace= True)
after= len(df)

print(before)
print(after)
```

```python
#Add column: numeric_churn
df['numeric_churn']= df['churn'].map({'Yes': 1, 'No': 0})

print(df['numeric_churn'].value_counts().to_string())
```

```python
#Add column: tenure_group
#Segment customers into 3 groups based on how long they have stayed

def assign_tenure_group(months):
    if months <= 12:
        return "New (0-12 mo)"
    elif months <= 36:
        return "Mid-term (13-36 mo)"
    else:
        return "Long-term (37+ mo)"

df['tenure_group']= df['tenure'].map(assign_tenure_group)

print(df['tenure_group'].value_counts().to_string())
```

```python
#Final data check
print(f"  Rows              : {df.shape[0]}")
print(f"  Columns           : {df.shape[1]}")
print(f"  Total null values : {df.isnull().sum().sum()}")
print(f"\n  Column dtypes after cleaning:")
print(df.dtypes.to_string())
```

```python
#Save the cleaned dataset

df.to_csv(r"C:\MY DOCUMENTS\Saiket System Projects\Cleaned_Telco_Customer_Churn.csv", index= False)
```
